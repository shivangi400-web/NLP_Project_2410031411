```python
import re

def analyze(text):

    patterns = [
        ("Currency", r'(?:₹|Rs\.?|INR|USD|\$|€|£)\s*\d+(?:,\d{3})*(?:\.\d+)?(?:\s*(?:crore|lakh|million|billion))?'),
        ("Percentage", r'\b\d+(?:\.\d+)?\s*%'),
        ("Ranking", r'\b\d+(?:st|nd|rd|th)\b|\b\d+(?:\.\d+)?\s*/\s*\d+(?:\.\d+)?'),
        ("Measurement", r'\b\d+(?:\.\d+)?\s*(?:kg|g|mg|km|m|cm|mm|ml|L|GB|MB|TB|GHz|MHz|Hz|inch|inches|°C|°F)\b'),
        ("Date", r'\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b|\b\d{1,2}\s+(?:Jan|January|Feb|February|Mar|March|Apr|April|May|June|Jun|July|Jul|August|Aug|September|Sep|October|Oct|November|Nov|December|Dec)\s+\d{4}\b'),
        ("Quantity", r'\b\d+(?:,\d{3})*(?:\.\d+)?\s*(?:million|billion|thousand)?\s*(?:users|students|employees|people|items|units|countries|customers|products|visitors|tickets|records|books|orders|files|pages|samples)?\b')
    ]

    results = []
    used = set()

    for category, pattern in patterns:

        for match in re.finditer(pattern, text, re.I):

            expr = match.group().strip()
            key = (match.start(), match.end())

            if key not in used:
                results.append((expr, category))
                used.add(key)

    print("\n" + "=" * 50)
    print("        NUMILEX - NLP ANALYSIS")
    print("=" * 50)

    if not results:
        print("No numerical expressions found.")
        return

    for expr, category in results:
        print(f"{expr:20} → {category}")

    print("=" * 50)
    print("Total Expressions:", len(results))

    print("\nCategory Summary:")

    for category in [
        "Quantity",
        "Percentage",
        "Currency",
        "Measurement",
        "Ranking",
        "Date"
    ]:
        count = sum(
            1 for _, c in results
            if c == category
        )
        print(f"{category:15}: {count}")


# ---------------------------------
# TEST YOUR NLP SYSTEM
# ---------------------------------

text = """
The company generated ₹12.5 crore revenue in 2025.
It achieved 18.5% growth and hired 500 employees.
The company expanded to 12 countries.
The server operates at 3.2 GHz.
The product received a 4.5/5 rating on 15 August 2025.
The laptop costs $1,299 and weighs 2.5 kg.
"""


analyze(text)
```
